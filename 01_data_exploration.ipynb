```python
import random

random.seed(42)

supplier_data = {
    "supplier_id": [f"SUP{i:03d}" for i in range(1, 11)],
    "supplier_name": [
        "Global Supply Co.",
        "Prime Logistics Suppliers",
        "Reliable Goods Ltd.",
        "Alpha Distribution",
        "BlueSky Suppliers",
        "NextGen Supply",
        "Trusted Trade Partners",
        "Metro Wholesale",
        "Swift Source Ltd.",
        "United Product Suppliers"
    ],
    "lead_time_days": [5, 7, 10, 6, 8, 12, 4, 9, 11, 7],
    "supplier_status": ["Active"] * 10
}

suppliers = pd.DataFrame(supplier_data)

display(suppliers)
```

```python
supplier_file = processed_folder / "suppliers.csv"

suppliers.to_csv(supplier_file, index=False)

print("Supplier data saved successfully!")
print("File location:", supplier_file)
```

```python
products = products.copy()

products["supplier_id"] = [
    random.choice(suppliers["supplier_id"].tolist())
    for _ in range(len(products))
]

product_supplier = products.merge(
    suppliers,
    on="supplier_id",
    how="left"
)

print("Total products:", len(product_supplier))
print("\nFirst 10 products with suppliers:")

display(product_supplier.head(10))
```

```python
product_supplier_file = processed_folder / "product_supplier_mapping.csv"

product_supplier.to_csv(product_supplier_file, index=False)

print("Product-supplier mapping saved successfully!")
print("Total mapped products:", len(product_supplier))
print("File location:", product_supplier_file)
```

```python
# Calculate the number of days in the historical dataset
total_days = (
    df_project["order_date"].max().date()
    - df_project["order_date"].min().date()
).days + 1

print("Historical period in days:", total_days)

# Calculate average daily demand for each product
product_demand["average_daily_demand"] = (
    product_demand["quantity"] / total_days
)

print("\nTop 10 products with average daily demand:")

display(
    product_demand[
        ["product_id", "product_name", "quantity", "average_daily_demand"]
    ].head(10)
)
```

```python
import numpy as np

np.random.seed(42)

# Connect demand information with supplier information
inventory = product_supplier.merge(
    product_demand[
        ["product_id", "quantity", "average_daily_demand"]
    ],
    on="product_id",
    how="left"
)

# Calculate safety stock (7 days of demand)
inventory["safety_stock"] = np.ceil(
    inventory["average_daily_demand"] * 7
).astype(int)

# Calculate reorder point
inventory["reorder_point"] = np.ceil(
    inventory["average_daily_demand"]
    * inventory["lead_time_days"]
    + inventory["safety_stock"]
).astype(int)

# Generate simulated current stock
inventory["current_stock"] = np.random.randint(
    100, 5001, size=len(inventory)
)

# Calculate days of inventory remaining
inventory["days_of_inventory"] = (
    inventory["current_stock"]
    / inventory["average_daily_demand"]
).round(2)

print("Simulated inventory created!")
print("Total products:", len(inventory))

display(
    inventory[
        [
            "product_id",
            "product_name",
            "supplier_id",
            "average_daily_demand",
            "lead_time_days",
            "safety_stock",
            "reorder_point",
            "current_stock",
            "days_of_inventory"
        ]
    ].head(10)
)
```

```python
# Generate a random stock coverage period between 3 and 60 days
simulated_stock_days = np.random.randint(
    3, 61, size=len(inventory)
)

# Calculate stock based on average daily demand
inventory["current_stock"] = np.ceil(
    inventory["average_daily_demand"] * simulated_stock_days
).astype(int)

# Ensure every product has at least 1 unit
inventory["current_stock"] = inventory["current_stock"].clip(lower=1)

# Recalculate days of inventory
inventory["days_of_inventory"] = (
    inventory["current_stock"]
    / inventory["average_daily_demand"]
).round(2)

print("Inventory quantities improved successfully!")

display(
    inventory[
        [
            "product_id",
            "product_name",
            "average_daily_demand",
            "lead_time_days",
            "safety_stock",
            "reorder_point",
            "current_stock",
            "days_of_inventory"
        ]
    ].head(10)
)
```

```python
inventory_file = processed_folder / "inventory.csv"

inventory.to_csv(inventory_file, index=False)

print("Inventory data saved successfully!")
print("Total inventory records:", len(inventory))
print("File location:", inventory_file)
```
